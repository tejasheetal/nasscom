```python
#2. Setup + Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Employee_ID": [101,102,103,104,105,106,107,108,109,110,
                    111,112,113,114,115],

    "Department": ["IT","HR","IT","Finance","HR",
                   "IT","Finance","IT","HR","Finance",
                   "IT","HR","Finance","IT","HR"],

    "Age": [22,25,28,30,24,35,40,27,29,45,32,26,38,31,50],

    "Experience": [1,2,4,6,2,10,15,3,5,18,8,3,12,7,20],

    "Salary": [25000,28000,32000,40000,27000,
               55000,80000,35000,38000,95000,
               60000,30000,75000,50000,150000],

    "Performance_Score": [65,70,75,80,68,85,90,78,72,95,
                          88,74,92,82,96],

    "Training_Completed": ["Yes","No","Yes","Yes","No",
                           "Yes","Yes","No","Yes","Yes",
                           "Yes","No","Yes","Yes","No"]
}

df = pd.DataFrame(data)

print(df)
```

```python
#3. Part A – Understanding Data Types
print("First 5 records:")
print(df.head())

print("\nLast 5 records:")
print(df.tail())

print("\nShape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nNumerical columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical columns:")
print(df.select_dtypes(exclude=np.number).columns.tolist())
```

```python
4. #Part B – Basic Data Exploration
print("Descriptive Statistics:")
print(df.describe())

print("\nEmployees in each department:")
print(df["Department"].value_counts())

print("\nTraining completed:")
print(df["Training_Completed"].value_counts())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
#5. Part C – Measures of Central Tendency
columns = ["Age", "Experience", "Salary", "Performance_Score"]

for column in columns:
    print("\n", column)
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
```

```python
#6. Part D – Measures of Spread / Dispersion
for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    print("\n", column)
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
    print("Range:", df[column].max() - df[column].min())
    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
```

```python
#7. Part E – Skewness
plt.hist(df["Salary"], bins=6, edgecolor="black")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()
```

```python
#8. Part F – Visualization
plt.hist(df["Salary"], bins=6, edgecolor="black")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()
```

```python
#Salary Box Plot
plt.boxplot(df["Salary"])
plt.ylabel("Salary")
plt.title("Salary Box Plot")
plt.show()
```

```python
#9. Part G – Combined Statistical Summary
summary = pd.DataFrame({
    "Mean": [
        df["Age"].mean(),
        df["Experience"].mean(),
        df["Salary"].mean(),
        df["Performance_Score"].mean()
    ],

    "Median": [
        df["Age"].median(),
        df["Experience"].median(),
        df["Salary"].median(),
        df["Performance_Score"].median()
    ],

    "Std_Deviation": [
        df["Age"].std(),
        df["Experience"].std(),
        df["Salary"].std(),
        df["Performance_Score"].std()
    ],

    "Skewness": [
        df["Age"].skew(),
        df["Experience"].skew(),
        df["Salary"].skew(),
        df["Performance_Score"].skew()
    ]
})

summary.index = ["Age", "Experience", "Salary", "Performance Score"]

print(summary)
```

```python
#10. Part H – Data Science Interpretation
df_new = df[df["Salary"] != 150000]

print("Original Mean:", df["Salary"].mean())
print("New Mean:", df_new["Salary"].mean())

print("\nOriginal Median:", df["Salary"].median())
print("New Median:", df_new["Salary"].median())

print("\nOriginal Standard Deviation:", df["Salary"].std())
print("New Standard Deviation:", df_new["Salary"].std())

print("\nOriginal Skewness:", df["Salary"].skew())
print("New Skewness:", df_new["Salary"].skew())
```

```python
#11. Part I – Group-Based Analysis
print("Mean Salary by Department:")
print(df.groupby("Department")["Salary"].mean())

print("\nMedian Salary by Department:")
print(df.groupby("Department")["Salary"].median())

print("\nStandard Deviation by Department:")
print(df.groupby("Department")["Salary"].std())
```

```python
#12. Part J – Final Challenge
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    print("\n", column)
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
    print("Range:", df[column].max() - df[column].min())
    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())
    print("Q1:", df[column].quantile(0.25))
    print("Q3:", df[column].quantile(0.75))
    print("IQR:", df[column].quantile(0.75) -
          df[column].quantile(0.25))
    print("Skewness:", df[column].skew())
```
