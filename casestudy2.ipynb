```python
#part-I
import numpy as np

import pandas as pd
import numpy as np

# Create the dataset
data = {
    'Student': ['S1','S2','S3','S4','S5','S6','S7','S8','S9','S10'],
    'Study_Hours': [2,3,4,5,6,7,8,4,6,9],
    'Attendance': [65,70,75,80,85,90,92,78,88,95],
    'Assignment_Score': [55,60,65,72,78,85,88,70,80,92],
    'Previous_Score': [58,62,68,70,76,82,86,72,79,90],
    'Final_Score': [60,64,70,75,82,88,91,74,84,94]
}

df = pd.DataFrame(data)

# Create X
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

# Create Y
Y = df["Final_Score"].values

# Add column of 1s for intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate coefficients
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```

**Students should explain, in simple words, what the estimated coefficients indicate about the relationship
between the input variables and final score.**

The coefficients indicate the relationship between the input variables and the Final Score. Study Hours has a negative coefficient, while Attendance and Assignment Score have positive coefficients. This means that, when the other variables are held constant, higher attendance and assignment scores are associated with higher predicted final scores. The negative Study Hours coefficient indicates a negative relationship in this fitted model.
