```python
#tips dataset
import seaborn as sns

tips = sns.load_dataset("tips")

tips.head()
```

```python
#Histogram
import matplotlib.pyplot as plt
plt.hist(

tips["total_bill"],

bins=20,

edgecolor="black"

)
```

```python
#pie chart
import matplotlib.pyplot as plt
day_counts=tips["day"].value_counts()
plt.pie(

day_counts,

labels=day_counts.index,

autopct="%1.1f%%",

startangle=90

)
```

```python
#scatter plot
import matplotlib.pyplot as plt
plt.scatter(
    tips["total_bill"],
    tips["tip"]
)
plt.xlabel("total bill")
plt.ylabel("tip")
plt.title("total bill vs tip")
plt.show()
```

```python
#bar plot
import matplotlib.pyplot as plt
avg_tip=tips.groupby("day")["tip"].mean()
plt.bar(
    avg_tip.index,
    avg_tip.values
)
plt.xlabel("day")
plt.ylabel("average tip")
plt.title("average tip by day")
plt.show()
```

```python
#horizantal bars
import matplotlib.pyplot as plt
avg_tip=tips.groupby("day")["tip"].mean()
plt.barh(
    avg_tip.index,
    avg_tip.values
)
plt.xlabel("average tip")
plt.ylabel("day")
plt.title("average tip by day")
plt.show()
```

```python
#box plot
import matplotlib.pyplot as plt
plt.boxplot(tips["total_bill"])
plt.ylabel("distribution of total bills")
plt.title("distribution of total bills")
plt.show()
```

```python
#count plot
import seaborn as sns
import matplotlib.pyplot as plt
sns.countplot(x="day", data=tips)
plt.xlabel("day")
plt.ylabel("count")
plt.title("number of customers by day")
plt.show()
```

```python
#swarm plot
import seaborn as sns
import matplotlib.pyplot as plt
sns.swarmplot(x="day", y="total_bill",data=tips)
plt.xlabel("day")
plt.ylabel("total_bill")
plt.title("total bill by day")
plt.show()
```

```python
#“Create a scatter plot of total bill vs tip, using smoker status to differentiate the points.”
import seaborn as sns
import matplotlib.pyplot as plt
sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="smoker",
    data=tips
)
plt.xlabel("total bill")
plt.ylabel("tip")
plt.title("total bill vs tip by smoker status")
plt.show()
```
